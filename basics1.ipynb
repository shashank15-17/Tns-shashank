### Conditional Program Example

This program demonstrates how to use `if`, `elif` (else if), and `else` statements to execute different blocks of code based on certain conditions. We'll check if a number is positive, negative, or zero.

```python
print("--- Simple Interactive Application ---")
print("Enter a number to check if it's positive, negative, or zero. Type 'quit' to exit.")

while True:
    user_input = input("Enter a number or 'quit': ")

    if user_input.lower() == 'quit':
        print("Exiting application. Goodbye!")
        break
    else:
        try:
            number = float(user_input) # Use float to handle decimals
            if number > 0:
                print(f"The number {number} is positive.")
            elif number < 0:
                print(f"The number {number} is negative.")
            else:
                print(f"The number {number} is zero.")
        except ValueError:
            print(f"Invalid input: '{user_input}'. Please enter a valid number or 'quit'.")

print("--- Application Finished ---")
```

### Listing Built-in Functions and Methods

Python provides several ways to inspect available functions and methods. Here, we'll demonstrate how to:
1.  List Python's built-in functions.
2.  List methods available for a specific object (e.g., a string).

```python
import inspect

# 1. Listing Python's built-in functions
print("--- Built-in Functions ---")
# Use dir(__builtins__) to get all built-in names
# Then filter for callables that are actual functions (not types or other objects)
built_in_functions = [name for name in dir(__builtins__) if inspect.isfunction(getattr(__builtins__, name, None))]

print(f"Total built-in functions: {len(built_in_functions)}")
print("Some examples:")
for func_name in sorted(built_in_functions)[:10]: # Displaying first 10 for brevity
    print(f"- {func_name}")


# 2. Listing methods of a string object
print("\n--- Methods of a String Object ---")
sample_string = "hello world"

# Use dir() to get all attributes and methods of an object
string_attributes = dir(sample_string)

# Filter for methods (callable attributes that don't start with '__')
string_methods = [attr for attr in string_attributes if callable(getattr(sample_string, attr)) and not attr.startswith('__')]

print(f"String object methods: {len(string_methods)}")
print("Some examples:")
for method_name in sorted(string_methods)[:10]: # Displaying first 10 for brevity
    print(f"- {method_name}")
```

### Functions within Modules (Classes) and Methods

This example demonstrates how to:
1.  Define a class, which acts as a logical grouping (like a module).
2.  Define a method within that class.
3.  Define a *nested function* inside that method.

This structure helps organize code and manage scope.

```python
class DataProcessorModule:
    """
    A class-based module to keep functions organized.
    It contains standard methods and a nested function example.
    """

    def __init__(self, data):
        self.data = data

    def process(self):
        """
        A method that performs processing using an internal nested function.
        """
        print(f"Processing data: {self.data}")

        def _internal_helper(x):
            """A function kept inside the method for internal logic."""
            return x.upper() if isinstance(x, str) else x * 2

        result = _internal_helper(self.data)
        print(f"Result: {result}")
        return result

    def display_info(self):
        """Another method inside the class module."""
        print("This is a method available within the DataProcessorModule.")

# Instantiate the 'module'
app = DataProcessorModule("hello world")

# Call the method that uses an internal function
app.process()

# Access another method in the module
app.display_info()
```
